```python
import tkinter as tk
from tkinter import simpledialog
from moviepy import VideoFileClip, TextClip, CompositeVideoClip
import whisper
```

```python
font_path = r"C:\Windows\Fonts\meiryo.ttc"  # パス自体は手元に保持（他で使う場合のため）

def show_text_editor(initial_text):
    result = {"text" : initial_text}
    window = tk.Tk()
    window.title("全原稿の確認・修正")

    # ラベルのフォント指定を修正（"Meiryo" または "メイリオ" を指定）
    label = tk.Label(window, text="AIが聞き取った全原稿です。自由に修正して完了を押してください",
                     font=("Meiryo", 11))
    label.pack(pady=10)

    text_box = tk.Text(window, wrap=tk.WORD, font=("Meiryo", 12))
    text_box.insert(tk.END, initial_text) # ※初期テキストを表示させるコードを追加しています
    text_box.pack(expand=True, fill=tk.BOTH, padx=10, pady=5)

    # 関数名を修正
    def on_submit():
        result["text"] = text_box.get("1.0", tk.END).strip()
        window.destroy()
    
    # commandの指定と、fontの指定方法を修正
    submit_btn = tk.Button(window, text="修正を完了して動画を作成する",
                           command=on_submit,
                           bg="#4CAF50", fg="white",
                           font=("Meiryo", 11, "bold")) # 太字にして見やすくしました
    submit_btn.pack(pady=10)
    
    window.mainloop()
    return result["text"]
```

```python
print("AIモデルを読み込んでいます")
model = whisper.load_model("base")
```

```python
print("動画の音声を解析中")
result = model.transcribe("Movie.MOV",language="ja")
detected_text = result["text"]
```

```python
corrected_text = show_text_editor(detected_text)
```

```python
print(f"最終決定されたテロップ:{corrected_text}")
```

```python
video = VideoFileClip("Movie.MOV")
text_clip = TextClip(
    text = corrected_text,
    font_size = 30,
    color = "white",
    font= font_path
)
```

```python
text_clip = text_clip.with_position(("center","bottom")).with_duration(16)
final_video = CompositeVideoClip([video, text_clip])
final_video.write_videofile("long_text_output.mp4")

print("長文対応のテロップ入り道が完成しました！")
```
