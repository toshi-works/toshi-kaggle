```python
from pydub import AudioSegment
import speech_recognition as sr

# 1. iPhoneのボイスメモ（.m4a）を自動で .wav に変換
print("① 音声ファイルを変換しています...")
audio = AudioSegment.from_file("sample.m4a", format="m4a")
audio.export("sample.wav", format="wav")
print("   -> 変換が完了しました。")

# 2. 音声認識の司令塔（耳）を準備
recognizer = sr.Recognizer()

# 3. 変換した .wav ファイルをプログラムに読み込ませる
print("② 音声データを解析中...")
with sr.AudioFile("sample.wav") as source:
    audio_data = recognizer.record(source) # 音声を丸ごとキャプチャ

# 4. Googleの音声認識サーバーに送って文字起こし
try:
    print("③ 文字起こしを実行しています...")
    text = recognizer.recognize_google(audio_data, language="ja-JP")
    
    print("\n====== 【文字起こし結果】 ======")
    print(text)
    print("=================================")

except sr.UnknownValueError:
    print("\n【エラー】音声がうまく聞き取れませんでした。もう少しハキハキ話した音声で試してみてください。")
except sr.RequestError:
    print("\n【エラー】インターネット接続に問題があるか、音声認識サービスが一時的に利用できません。")
```
