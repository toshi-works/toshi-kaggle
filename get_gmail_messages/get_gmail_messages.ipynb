```python
import base64
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build

# 1. Googleの機能（API）を使うための権限を設定
# ※今回はメールを「読み込む（readonly）」だけの安全な設定です
SCOPES = ['https://www.googleapis.com/auth/gmail.readonly']

def get_gmail_messages():
    # 認証処理（ブラウザが開いて「アクセスを許可しますか？」と聞かれます）
    flow = InstalledAppFlow.from_client_secrets_file('credentials.json', SCOPES)
    creds = flow.run_local_server(port=0)
    
    # Gmailを操作するサービスを起動
    service = build('gmail', 'v1', credentials=creds)
    
    # 2. 直近の受信トレイから最新の「5件」のメールIDを取得
    results = service.users().messages().list(userId='me', maxResults=5, labelIds=['INBOX']).execute()
    messages = results.get('messages', [])
    
    if not messages:
        print('メールが見つかりませんでした。')
        return

    print(f"--- 最新のメールを {len(messages)} 件取得しました ---")
    
    # 3. 取得したIDを元に、メールの「中身（件名や本文）」を詳しく読み解く
    for msg in messages:
        txt = service.users().messages().get(userId='me', id=msg['id'], format='raw').execute()
        msg_str = base64.urlsafe_b64decode(txt['raw'].encode('ASCII'))
        
        # ※ここでは簡易的にメールのIDとデータの一部を表示します
        print(f"メールID: {msg['id']}")
        # ここに今後、件名や本文をきれいに抜き出すコードを追加していきます

# 実行する
if __name__ == '__main__':
    get_gmail_messages()
```
