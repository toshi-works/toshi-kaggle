```python
import pandas as pd
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
```

```python
pd.set_option('display.max_columns', None)
train_df.head()
```

```python
def year_of_construction(df):
    df['year_of_construction'] = df['YrSold'] - df['YearBuilt']
    return df
def all_porch(df):
    df['all_porch'] = df['OpenPorchSF'] + df['EnclosedPorch'] + df['3SsnPorch'] + df['ScreenPorch']
    return df  
def average_rooms(df):
    df['average_rooms'] = df['GrLivArea'] / df['TotRmsAbvGrd']
    return df
def Building_Coverage_Ratio(df):
    df['building_coverage_ratio'] = df['GrLivArea'] / df['LotArea']
    return df
def total_square_feet(df):
    df['TotalSF'] = df['1stFlrSF'] + df['2ndFlrSF'] + df['TotalBsmtSF']
    return df
```

```python
train_df = year_of_construction(train_df)
test_df = year_of_construction(test_df)
train_df = all_porch(train_df)
test_df = all_porch(test_df)
train_df = average_rooms(train_df)
test_df = average_rooms(test_df)
train_df = Building_Coverage_Ratio(train_df)
test_df = Building_Coverage_Ratio(test_df)
train_df = total_square_feet(train_df)
test_df = total_square_feet(test_df)
```

```python
train_df.head()
```

```python
train_df.isnull().sum().sort_values(ascending=False).head(20)
```

```python
# 数値（int型やfloat型）の列だけを選んで、Nullを0で埋める
y_train = train_df['SalePrice']
train_df.drop('SalePrice', axis=1, inplace=True)
num_cols = train_df.select_dtypes(include=['number']).columns
train_df[num_cols] = train_df[num_cols].fillna(0)
test_df[num_cols] = test_df[num_cols].fillna(0)
# 文字列（object型）の列だけを選んで、Nullを'None'で埋める
str_cols = train_df.select_dtypes(include=['object']).columns
train_df[str_cols] = train_df[str_cols].fillna('None')
test_df[str_cols] = test_df[str_cols].fillna('None')
```

```python
train_df.isnull().sum().sum()
X_train = train_df
X_test = test_df
# 1. 訓練データの中から「地上階が4000超」かつ「価格が30万ドル未満」の異常な大豪邸のインデックスを見つける
#outliers = X_train[(X_train['GrLivArea'] > 4000) & (y_train < 300000)].index
# 2. その2件を、訓練データ（X_train, y_train）から完全に削除（drop）する
# X_train = X_train.drop(outliers)
# y_train = y_train.drop(outliers)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 2行2列の、合計4つのグラフを並べるキャンバスを作る
fig, axes = plt.subplots(1, 4, figsize=(6, 5))

# それぞれの場所に箱ひげ図を割り当てる
sns.boxplot(y=X_train['GrLivArea'], ax=axes[0])
sns.boxplot(y=X_train['TotalBsmtSF'], ax=axes[1])
sns.boxplot(y=X_train['LotArea'], ax=axes[2])
sns.boxplot(y=X_train['TotalSF'], ax=axes[3])

plt.tight_layout() # グラフ同士の被りを自動で調整
plt.show()
```

```python
# 縦横をひっくり返して表示する
print(X_train.describe().T)
```

```python
import numpy as np
# 1. trainとtestの形を揃えるために、一度縦に結合する
all_df = pd.concat([X_train, X_test], axis=0)
# 1. 数値列（int型、float型）の列名だけをリストで取得する
num_cols = all_df.select_dtypes(include=['number']).columns
# 2. 各列の歪度（skew）を計算し、歪度が 0.75 以上の列だけをあぶり出す
skewed_cols = all_df[num_cols].skew()[all_df[num_cols].skew() > 1.5].index
all_df[skewed_cols] = np.log1p(all_df[skewed_cols])
# MSSubClassを数値型（int）から文字列型（str/object）に変換する
all_df['MSSubClass'] = all_df['MSSubClass'].astype(str)
# この状態で get_dummies を使えば、MSSubClass もちゃんと狙い撃ち（展開）してくれます！
all_df_encoded = pd.get_dummies(all_df)
# 2. 結合したデータに対して、一括でOne-Hot Encodingを行う
all_df_encoded = pd.get_dummies(all_df)
# 3. エンコーディングが終わったら、また元の train の行数と test の行数に切り分ける
# (元の X_train の行数が 1460行 だとしたら、先頭から1460行を X_train に戻す)
X_train_encoded = all_df_encoded.iloc[:len(X_train)]
X_test_encoded = all_df_encoded.iloc[len(X_train):]
```

```python
all_df_encoded.shape
```

```python
# 必要な道具（モデル）をインポートする
from sklearn.ensemble import RandomForestRegressor
# モデルの「箱」を作る（初期化）
# random_state=42 は、毎回同じ結果になるようにするための「乱数の固定」です。
model = RandomForestRegressor(random_state=42)
# モデルに学習させる
model.fit(X_train_encoded, y_train)
# テストデータの予測値を計算する
predictions = model.predict(X_test_encoded)
```

```python
# 提出サンプルのファイルを読み込む
sub = pd.read_csv('sample_submission.csv')
# サンプルの価格列を、自分が予測したデータに上書きする
sub['SalePrice'] = predictions
# テストデータのIDと、提出ファイルのIDが全員一致しているかチェックする（Trueなら安心！）
print((X_test['Id'] == sub['Id']).all())
# CSVファイルとして保存する
# index=False をつけないと、余計な行番号が保存されてしまう
sub.to_csv('my_submission.csv', index=False)
```

```python
import numpy as np

# モデルから重要度を取り出して、列名とセットにする
importances = model.feature_importances_
column_names = X_train_encoded.columns

# 見やすいようにデータフレームにまとめる
feature_imp_df = pd.DataFrame({
    'Feature': column_names,
    'Importance': importances
})

# 重要度が高い順（降順）に並び替えて、トップ10を表示する
# 💡【クイズ・ヒント】並び替えるPandasの関数は何だったでしょうか？
# feature_imp_df = feature_imp_df.xxxx_values(by='Importance', ascending=False)

print(feature_imp_df.sort_values(by='Importance', ascending=False).head(10))
```

```python
import lightgbm as lgb

# 1. モデルの初期化
lgb_model = lgb.LGBMRegressor(n_estimators=100, random_state=42)

# 2. 学習させる
lgb_model.fit(X_train_encoded, y_train)

# 3. テストデータの予測をする
lgb_predictions = lgb_model.predict(X_test_encoded)
```

```python
# 提出サンプルを読み込んで、LightGBMの予測値で上書き
sub = pd.read_csv('sample_submission.csv')
sub['SalePrice'] = lgb_predictions

# 新しいファイル名で保存
sub.to_csv('lgb_submission.csv', index=False)
```

```python
import numpy as np

# 1. 学習させる前に、ターゲット（SalePrice）をログ変換する！
y_train_log = np.log1p(y_train)

# 2. ログ変換したターゲットを使って、LightGBMに学習させる
lgb_model = lgb.LGBMRegressor(n_estimators=100, random_state=42)
lgb_model.fit(X_train_encoded, y_train_log)  # 💡ここでy_train_logを渡す

# 3. テストデータを予測する（※この予測値はまだ「ログの世界」の数値です）
lgb_predictions_log = lgb_model.predict(X_test_encoded)

# 4. 【超重要】元の「ドル」の単位に逆変換して戻す！
final_predictions = np.expm1(lgb_predictions_log)
```

```python
# 提出サンプルを読み込んで、LightGBMの予測値で上書き
sub = pd.read_csv('sample_submission.csv')
sub['SalePrice'] = final_predictions

# 新しいファイル名で保存
sub.to_csv('lgb_submission.csv', index=False)
```

```python
# モデルの箱を作る部分を書き換えます
# 慎重に（learning_rate=0.05）、じっくりたくさん（n_estimators=500）学習させます
lgb_model = lgb.LGBMRegressor(
    n_estimators=500, 
    learning_rate=0.05, 
    random_state=42
)

# あとの流れ（fit、predict、逆変換、CSV保存）は全く同じでOKです！
lgb_model.fit(X_train_encoded, y_train_log)
lgb_predictions_log = lgb_model.predict(X_test_encoded)
final_predictions = np.expm1(lgb_predictions_log)

# 保存して提出ファイルを作成
sub = pd.read_csv('sample_submission.csv')
sub['SalePrice'] = final_predictions
sub.to_csv('lgb_tuned_submission.csv', index=False)
```
