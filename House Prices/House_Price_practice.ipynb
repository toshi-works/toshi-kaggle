```python
import pandas as pd
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
```

```python
pd.set_option('display.max_columns', None)
train_df.head()
```

```python
def year_of_construction(df):
    df['year_of_construction'] = df['YrSold'] - df['YearBuilt']
    return df
def all_porch(df):
    df['all_porch'] = df['OpenPorchSF'] + df['EnclosedPorch'] + df['3SsnPorch'] + df['ScreenPorch']
    return df  
def average_rooms(df):
    df['average_rooms'] = df['GrLivArea'] / df['TotRmsAbvGrd']
    return df
def Building_Coverage_Ratio(df):
    df['building_coverage_ratio'] = df['GrLivArea'] / df['LotArea']
    return df
```

```python
train_df = year_of_construction(train_df)
test_df = year_of_construction(test_df)
train_df = all_porch(train_df)
test_df = all_porch(test_df)
train_df = average_rooms(train_df)
test_df = average_rooms(test_df)
train_df = Building_Coverage_Ratio(train_df)
test_df = Building_Coverage_Ratio(test_df)
```

```python
train_df.head()
```

```python
train_df.isnull().sum().sort_values(ascending=False).head(20)
```

```python
# 数値（int型やfloat型）の列だけを選んで、Nullを0で埋める
y_train = train_df['SalePrice']
train_df.drop('SalePrice', axis=1, inplace=True)
num_cols = train_df.select_dtypes(include=['number']).columns
train_df[num_cols] = train_df[num_cols].fillna(0)
test_df[num_cols] = test_df[num_cols].fillna(0)
# 文字列（object型）の列だけを選んで、Nullを'None'で埋める
str_cols = train_df.select_dtypes(include=['object']).columns
train_df[str_cols] = train_df[str_cols].fillna('None')
test_df[str_cols] = test_df[str_cols].fillna('None')
```

```python
train_df.isnull().sum().sum()
X_train = train_df
X_test = test_df
```

```python
# 1. trainとtestの形を揃えるために、一度縦に結合する
all_df = pd.concat([X_train, X_test], axis=0)
# MSSubClassを数値型（int）から文字列型（str/object）に変換する
all_df['MSSubClass'] = all_df['MSSubClass'].astype(str)
# この状態で get_dummies を使えば、MSSubClass もちゃんと狙い撃ち（展開）してくれます！
all_df_encoded = pd.get_dummies(all_df)
# 2. 結合したデータに対して、一括でOne-Hot Encodingを行う
all_df_encoded = pd.get_dummies(all_df)
# 3. エンコーディングが終わったら、また元の train の行数と test の行数に切り分ける
# (元の X_train の行数が 1460行 だとしたら、先頭から1460行を X_train に戻す)
X_train_encoded = all_df_encoded.iloc[:len(X_train)]
X_test_encoded = all_df_encoded.iloc[len(X_train):]
```

```python
all_df_encoded.shape
```

```python
# 必要な道具（モデル）をインポートする
from sklearn.ensemble import RandomForestRegressor
# モデルの「箱」を作る（初期化）
# random_state=42 は、毎回同じ結果になるようにするための「乱数の固定」です。
model = RandomForestRegressor(random_state=42)
# モデルに学習させる
model.fit(X_train_encoded, y_train)
# テストデータの予測値を計算する
predictions = model.predict(X_test_encoded)
```

```python
# 提出サンプルのファイルを読み込む
sub = pd.read_csv('sample_submission.csv')
# サンプルの価格列を、自分が予測したデータに上書きする
sub['SalePrice'] = predictions
# テストデータのIDと、提出ファイルのIDが全員一致しているかチェックする（Trueなら安心！）
print((X_test['Id'] == sub['Id']).all())
# CSVファイルとして保存する
# index=False をつけないと、余計な行番号が保存されてしまう
sub.to_csv('my_submission.csv', index=False)
```
