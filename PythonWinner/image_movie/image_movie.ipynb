```python
import cv2

# 画像を読み込み
img = cv2.imread("Google.png")

# 画像のNumPy配列を表示
# print(img)

# 画像の行(高さ)、列(幅)、チャンネル数を表示
print(img.shape)
```

```python
# cv2.imshow("image", img)
# キーボード入力を待つ。何かキーが押されたら終了する
# cv2.waitKey(0)
```

```python
# 画像を "image05_2_2_1.jpg" という名前で保存
cv2.imwrite("google.jpg", img)
```

```python
# 画像を縮小して保存する
img_small = cv2.resize(img, (100, 50))
cv2.imwrite("google_small.jpg", img_small)

# 画像を2倍に拡大して保存する
img_large = cv2.resize(img, None, fx=0.5, fy=0.5)
cv2.imwrite("half_google.jpg", img_large)
```

```python
# 画像の行(高さ)、列(幅)、チャンネル数(BGR)の情報を個別に格納する
h, w, _ = img.shape
# 画像を切り取る
img_cropped = img[
    0 : round(h / 2), 0 : round(w / 2), :
]
# 切り取った画像をファイルに保存
cv2.imwrite("google_cropped.jpg", img_cropped)
```

```python
# 画像を切り取る
img_cropped2 = img[
    round(h * 0.25) : round(h * 0.75), round(w * 0.25) : round(w * 0.75), :
]

# 切り取った画像をファイルに保存
cv2.imwrite("google_cropped2.jpg", img_cropped2)
```

```python
# 画像の配列をコピーする
img_changed = img.copy()
# 画像の緑と青を0にする
img_changed[:, :, (0, 1)] = 0
# 変更した画像をファイルに保存
cv2.imwrite("google_red.jpg", img_changed)
```

```python
# 画像をグレースケールで読み込み
gray = cv2.imread("Google.png", cv2.IMREAD_GRAYSCALE)

# グレースケールとして読み込んだ画像をファイルに保存
cv2.imwrite("google_gray.jpg", gray)
```

```python
from cv2.bgsegm import createBackgroundSubtractorMOG

# 画像の読み込み
img1 = cv2.imread("sample_img02_01.jpg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("sample_img02_02.jpg", cv2.IMREAD_GRAYSCALE)
# 背景差分検出のためのオブジェクトの生成
model = createBackgroundSubtractorMOG()
# はじめに背景画像を与え、次に物体が追加された画像を与える
model.apply(img1)
diff = model.apply(img2)
# 差分 (マスク) をファイルに保存する
cv2.imwrite("diff.jpg", diff)
```
